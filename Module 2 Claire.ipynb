FSHD is a skeletal muscle disease that causes progressive muscle weakness due to a derepressed DUX4 gene. Individuals with FSHD often asymmetrically experience fat replacing their skeletal muscle. In our analysis, we want to answer the question of whether the fat fraction in the muscle is a valid way to measure the disease progression. We will be testing the fat fraction of both the left and the right side for each individual to ensure that the tests can be applied to both. This is important because it can allow healthcare providers to track the progression of FSHD and determine what treatments should be provided.

* Symptoms
  * An early and common symptom of FSHD is facial weakness. Other common symptoms include foot drop which causes trouble walking, shoulder weakness, abdominal and posture issues, hip weakness, and chronic pain. [5](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms)
  * Less prevalent symptoms of FSHD include joint/spinal issues, respiratory/heart involvement, retinal issues, hearing loss, and difficulty swallowing. FSHD is also often noted to be asymmetrical, affecting each side of the body differently. [5](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms)
* Diagnosis
  * Suspected FSHD typically starts with muscle weakness. At this point, the patient will be evaluated by a neuromuscular specialist or neurologist. The most reliable way to diagnose FSHD is through genetic testing. Genetic testing can identify a contraction of D4Z4 on the chromosome. In other cases, the patient may undergo blood testing to measure creatine kinase levels, an electromyography to analyze muscle activity, or a muscle biopsy. [6](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/diagnosis)

- The MRI examinations were conducted using 3T Siemens scanners and were focused in the lower leg area. 
- The researchers ran T1 Dixon scans through a deep-learning algorithm that detected and created boundaries of the tibialis anterior muscle for every slice. They also split the data into water signals and fat signals.

## Limitations and Future Work:
Claire:

Describe the limitations of your project, including limitations of the sample size, available visits, MRI measurements, segmentation, biomarkers, functional measures, and analysis methods.

If you had more time or additional data, explain how you would improve the analysis. Examples include analyzing more participants, adding more timepoints, improving segmentation, comparing additional muscles, validating the MRI measure against functional outcomes, or testing whether the measure is sensitive to treatment response.

The dataset had a small sample of 34 participants, so it may not be able to properly reflect the entire population of individuals with FSHD. Also, it is difficult to get people to come back at the exact 1 or 2 year mark, so we had to interpolate the data to make a guess of where their data values would be between each point. With three time points for the main dataset and two timepoints for the MRI samples, it's difficult to truly understand the progression of the disease. Additionally, the MRI information came from one patient, so it cannot necessarily be used to predict the images of other people. Segmentation by computers can also include errors in boundaries that might require human correction. Variability between biomarker levels and functional can also cause issues when analyzing data and they do not perfectly depict the development of FSHD as it could be an off day for the person. Methods like interpolation can also cause data points to be incorrect from the true value at the time interpolated.

If we had more data points, we could be more certain in our analysis because it would be reflective of a larger sample of people. More timepoints would also be helpful so we wouldn't have to make as many assumptions of the year or more between each visit. We also could have improved deep learning models that would be able to segment the MRI data more accurately. Additionally, having more MRI data for different types of muscles could also be useful in ensuring that our analysis is representing the whole body of the FSHD affected individual. With more time, we would compare using the fat fraction of the left side of the body too to make sure that the body is being equally tested.
