## Disease Background:

*The following 7 bullets seem most relevant, but feel free to choose others if you like:*

* Prevalence & incidence
  * The prevalence of FSHD is estimated to be around 4-12 out of 100,000 individuals. This number differs from country to country, with China having around 0.75 cases for million individuals and the Netherlands experiencing 12 per 100,000 individuals. [1](https://pmc.ncbi.nlm.nih.gov/articles/PMC13038818/) [2](https://www.fshdsociety.org/2014/09/01/fshd-one-prevalent-neuromuscular-disorders/)
  * The birth incidence of FSHD is around 1/7,500 individuals to 1/14,286 individuals across the world. [3](https://www.umassmed.edu/wellstone/overview/fshdfacts/) The average amount of individuals diagnosed with FSHD each year is 52 people every year, and the incidence rate in the Netherlands is about 0.3/100,000 person-years. [4](https://pmc.ncbi.nlm.nih.gov/articles/PMC4166358/)
* Symptoms
  * An early and common symptom of FSHD is facial weakness. Other common symptoms include foot drop which causes trouble walking, shoulder weakness, abdominal and posture issues, hip weakness, and chronic pain. [5](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms)
  * Less prevalent symptoms of FSHD include joint/spinal issues, respiratory/heart involvement, retinal issues, hearing loss, and difficulty swallowing. FSHD is also often noted to be asymmetrical, affecting each side of the body differently. [5](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms)
* Diagnosis
  * Suspected FSHD typically starts with muscle weakness. At this point, the patient will be evaluated by a neuromuscular specialist or neurologist. The most reliable way to diagnose FSHD is through genetic testing. Genetic testing can identify a contraction of D4Z4 on the chromosome. In other cases, the patient may undergo blood testing to measure creatine kinase levels, an electromyography to analyze muscle activity, or a muscle biopsy. [6](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/diagnosis)
* Standard of care treatments (& reimbursement) 
  * Since there is no current cure or specific treatment for FSHD, rather there is pain and symptom management, supportive care, and etc. To suppress the pain caused my musculoskeletal pain and fatigue many doctors reccomend going on medication, such as anti-depressants, antiepileptics, and ant-inflamatory drugs. Other non-medicated methods include stretching and heat therapy. Supportive care includes the use of anklefoot orthoses (AFOs) to stop potential falls and ensure stability when walking. And of course there is moderate to low intensity excercise to maintain strenth in areas that are affected less.  [7](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/medical-management)
* Disease progression & prognosis
  * Typically signs of FSHD start to appear from teenage years into the 30s, with 10% starting before the age of 10. Early on-set cases show signs of weakness in the face, in the scapula, or in the upper arms and slowly start to progress towards the lower body. By the age of 50, 20% will be requiring wheel chairs. And the impact of this disease has high variability, so no one person has the same affects. FSHD does not have a direct impact on the lifespan of those affected. Although, many suffer from severe depression and experiance mental health tolls from this disease. [8](https://www.fshdsociety.org/living-with-fshd/understanding-fshd/) [9](https://musculardystrophynews.com/muscular-dystrophy-types/facioscapulohumeral-muscular-dystrophy/)
* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
  * FSHD is a genetic disorder that primarily affects skeletal muscles, especially those in the face, shoulders, and upper arms. There are two types of FSHD: FSHD1 and FSHD2. They have different genetic causes but ultimately affect the same molecular pathway. FSHD1 accounts for approximately 95% of cases and occurs when the D4Z4 DNA repeat region on chromosome 4 becomes shortened from the normal 11–100 repeats to only 1–10 repeats. FSHD2 is less common and typically results from mutations in genes such as SMCHD1, which help regulate DNA activity.
  [10](https://www.ncbi.nlm.nih.gov/books/NBK1443/?report=reader)
* Clinical Trials/next-gen therapies
   * Currently, there are no approved disease-modifying treatments for FSHD, but researchers are developing therapies that target the underlying genetic mechanisms rather than just managing symptoms. One major approach involves reducing DUX4 expression through RNA-based therapies. Delpacibart braxlosiran (del-brax), an investigational treatment originally developed by Avidity Biosciences and now under Novartis, uses RNA-targeting technology to reduce DUX4 production in muscle cells. In June 2026, results from the Phase 1/2 FORTITUDE trial showed reductions in biomarkers associated with DUX4 activity and muscle damage. A Phase 3 trial is currently underway to further evaluate its effectiveness and safety. [11](https://www.novartis.com/us-en/news/media-releases/novartis-delpacibart-braxlosiran-del-brax-phase-iii-study-facioscapulohumeral-muscular-dystrophy-fshd-meets-primary-biomarker-endpoint)

The example MRI dataset includes eight cross-sectional images taken at different locations along the TA muscle during each of the two visits. The files are stored as NumPy arrays and can be loaded into Python for analysis. By comparing the fat signal to the combined water and fat signals at each pixel, we can calculate the fat fraction, which provides an estimate of the percentage of fat within the tissue.

- How the water, fat, and segmentation-mask files were loaded
    * Python was used to load the MRI data, with NumPy for numerical calculations, pandas for organizing the data, and Matplotlib for visualization. The slice_info.csv file contains information about each MRI slice, including its timepoint, location, and corresponding water, fat, and segmentation-mask files. We first selected a representative slice from TP1 and loaded the three corresponding .npy files using np.load(). The segmentation mask was converted into a Boolean array to distinguish pixels belonging to the tibialis anterior muscle from those outside it.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")

row = manifest[manifest["timepoint"] == "TP1"].iloc[3]

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

- How the tibialis anterior mask was used to identify the muscle
    * The segmentation mask was used to isolate the tibialis anterior muscle from the surrounding tissue in each MRI image. The mask identifies pixels belonging to the TA muscle with a value of True, while pixels outside the muscle are assigned False. After calculating fat fraction, we applied the mask by setting values outside the muscle to NaN. This ensures that surrounding fat, bone, and other tissues are excluded from the muscle-specific analysis. Using the mask allows us to calculate more representative fat-fraction measurements for the tibialis anterior rather than the entire MRI image.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")

row = manifest[manifest["timepoint"] == "TP1"].iloc[3]

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```


- How fat fraction was calculated for the whole image region and for the TA muscle only
