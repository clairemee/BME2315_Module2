- How the tibialis anterior mask was used to identify the muscle
    * The segmentation mask was used to isolate the tibialis anterior muscle from the surrounding tissue in each MRI image. The mask identifies pixels belonging to the TA muscle with a value of True, while pixels outside the muscle are assigned False. After calculating fat fraction, we applied the mask by setting values outside the muscle to NaN. This ensures that surrounding fat, bone, and other tissues are excluded from the muscle-specific analysis. Using the mask allows us to calculate more representative fat-fraction measurements for the tibialis anterior rather than the entire MRI image.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
# Use the dataset path already found in the notebook if available.
manifest_path = globals().get("manifest_path", DATA_DIR / "slice_info.csv")

if not Path(manifest_path).exists():
    matches = list(Path.cwd().rglob("FSHD_TA_MRI_data_0020_TP1_TP2/slice_info.csv"))
    if not matches:
        raise FileNotFoundError(
            f"Could not find slice_info.csv. Current working directory: {Path.cwd()}. "
            "Make sure the dataset folder is downloaded and extracted."
        )
    manifest_path = matches[0]
    DATA_DIR = manifest_path.parent

manifest = pd.read_csv(manifest_path)

row = manifest[manifest["timepoint"] == "TP1"].iloc[3]

DATA_DIR = Path(manifest_path).parent
water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```


- How fat fraction was calculated for the full image region and the TA muscle
    * Fat fraction was calculated at each pixel as (fat signal / (fat signal + water signal)) × 100. It first evaluated this ratio across all valid pixels in the supplied image region. It then applied the tibialis anterior (TA) segmentation mask to retain only pixels within the muscle, excluding surrounding tissues from the TA-specific analysis.
    

```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

- How the fat-fraction distribution was summarized
    * The distribution of fat-fraction values was visualized using a histogram and summarized with the mean, median, and standard deviation.



```python
# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.axis("off")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

plt.show()

```

## Interpolation Analysis:

Explain how interpolation was used to estimate fat fraction or another measurement at a location between sampled points along the tibialis anterior.

Include:

- The variable used to define location along the muscle, such as distance in millimeters or percentage of muscle length
    - Our interpolation used time since baseline, measured in months, as the independent variable. 
- The sampled locations used as inputs to the interpolation
    - The inputs were each participant’s actual baseline, Year 1, and Year 2 visit times, rather than sampled locations along the muscle. Linear interpolation used baseline and Year 1 measurements, while quadratic interpolation used all three visits.
- The interpolation method and Python code used

    - We used linear and quadratic interpolation to estimate right tibialis anterior fat fraction and clinical severity at common times after baseline. Linear interpolation used baseline and Year 1 measurements to estimate values at 11.15 months. Quadratic interpolation used all three visits to estimate values at 22.13 months. These times were the earliest recorded Year 1 and Year 2 visits in the dataset.

    - We used NumPy arrays to store visit times and measurements, then used SciPy’s interp1d function to calculate the estimates. In the analysis of all participants, missing estimates were recorded as NaN when the required measurements were unavailable or the target time was outside the measured range.

    - The following code shows participant 13-0004 as an example. Interpolated values are estimates between visits, rather than new measurements. This also applies to fractional clinical severity scores, which are mathematical estimates rather than actual clinical ratings.

### Fat fraction for Participant 13-0004

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

df = pd.read_csv("Module2_FSHD_Dataset.csv")

T1 = df["lag_months_y1"].min()
T2 = df["lag_months_y2"].min()

row = df.loc[df["subject_id"] == "13-0004"].iloc[0]

months = np.array([
    row["lag_months_baseline"],
    row["lag_months_y1"],
    row["lag_months_y2"]
])

# Fat fraction for the example participant
ff_values = np.array([
    row["ff_baseline_R"],
    row["ff_y1_R"],
    row["ff_y2_R"]
])

ff_linear = interp1d(months[:2], ff_values[:2], kind="linear")
ff_quadratic = interp1d(months, ff_values, kind="quadratic")

ff_T1 = float(ff_linear(T1))
ff_T2 = float(ff_quadratic(T2))

linear_months = np.linspace(months[0], months[1], 100)
quadratic_months = np.linspace(months[0], months[2], 200)

plt.figure(figsize=(8, 5))
plt.plot(
    linear_months, ff_linear(linear_months),
    color="#e0569b", linestyle="--",
    label="Linear interpolation: baseline to Year 1"
)
plt.plot(
    quadratic_months, ff_quadratic(quadratic_months),
    color="#e8821e", linestyle=":",
    label="Quadratic interpolation: all three visits"
)
plt.scatter(
    months, ff_values, color="#1f7a8c",
    s=90, zorder=4, label="Measured values"
)
plt.scatter(
    T1, ff_T1, color="#e0569b", marker="D", s=100,
    zorder=5, label=f"Estimate at {T1:.2f} months"
)
plt.scatter(
    T2, ff_T2, color="#e8821e", marker="D", s=100,
    zorder=5, label=f"Estimate at {T2:.2f} months"
)
plt.title("Right tibialis anterior fat fraction over time\nParticipant 13-0004")
plt.xlabel("Time since baseline (months)")
plt.ylabel("Right TA fat fraction (%)")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

print(f"Estimated fat fraction at {T1:.2f} months: {ff_T1:.2f}%")
print(f"Estimated fat fraction at {T2:.2f} months: {ff_T2:.2f}%")
```

### Clinical Severity for the same participant

```python
severity_values = np.array([
    row["clinical_severity_score_baseline"],
    row["clinical_severity_score_y1"],
    row["clinical_severity_score_y2"]
])

severity_linear = interp1d(
    months[:2], severity_values[:2], kind="linear"
)
severity_quadratic = interp1d(
    months, severity_values, kind="quadratic"
)

severity_T1 = float(severity_linear(T1))
severity_T2 = float(severity_quadratic(T2))

plt.figure(figsize=(8, 5))
plt.plot(
    linear_months, severity_linear(linear_months),
    color="#e0569b", linestyle="--",
    label="Linear interpolation: baseline to Year 1"
)
plt.plot(
    quadratic_months, severity_quadratic(quadratic_months),
    color="#e8821e", linestyle=":",
    label="Quadratic interpolation: all three visits"
)
plt.scatter(
    months, severity_values, color="#1f7a8c",
    s=90, zorder=4, label="Measured values"
)
plt.scatter(
    T1, severity_T1, color="#e0569b", marker="D", s=100,
    zorder=5, label=f"Estimate at {T1:.2f} months"
)
plt.scatter(
    T2, severity_T2, color="#e8821e", marker="D", s=100,
    zorder=5, label=f"Estimate at {T2:.2f} months"
)
plt.title("Clinical severity over time\nParticipant 13-0004")
plt.xlabel("Time since baseline (months)")
plt.ylabel("Clinical severity score (points)")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

print(f"Estimated severity at {T1:.2f} months: {severity_T1:.2f}")
print(f"Estimated severity at {T2:.2f} months: {severity_T2:.2f}")
```

### From Baseline to Year 1

```python
from scipy.stats import spearmanr

correlation_results = []
# Baseline to Year 1
paired = df[['subject_id', 'ff_baseline_R', 'ff_y1_R',
             'clinical_severity_score_baseline', 'clinical_severity_score_y1',
             'lag_months_baseline', 'lag_months_y1']].dropna().copy()

paired['ff_change'] = paired['ff_y1_R'] - paired['ff_baseline_R']
paired['severity_change'] = (
    paired['clinical_severity_score_y1'] - paired['clinical_severity_score_baseline']
)
n = len(paired)
rho = np.nan
p_approx = np.nan
p_shuffle = np.nan

# We need some variation in both changes to calculate a correlation
if n >= 3 and paired['ff_change'].nunique() > 1 and paired['severity_change'].nunique() > 1:
    rho, p_approx = spearmanr(paired['ff_change'], paired['severity_change'])

    # Shuffle the pairings to check the p-value with this small group
    # The seed makes the answer repeatable when someone else runs it
    rng = np.random.default_rng(42)
    extreme = 0
    for repeat in range(9999):
        shuffled = rng.permutation(paired['severity_change'].to_numpy())
        shuffled_rho, unused_p = spearmanr(paired['ff_change'], shuffled)
        if abs(shuffled_rho) >= abs(rho) - 1e-12:
            extreme = extreme + 1
    p_shuffle = (extreme + 1) / 10000
else:
    print('Not enough data or variation for the Year 1 correlation.')

print('\nBaseline to Year 1:')
print('Participants with both changes:', n)
print('Spearman rho:', rho)
print('Approximate p-value:', p_approx)
print('Two-sided shuffled p-value:', p_shuffle)
print('Follow-up months (minimum, median, maximum):',
      paired['lag_months_y1'].min(), paired['lag_months_y1'].median(), paired['lag_months_y1'].max())

correlation_results.append({
    'comparison': 'Baseline to Year 1',
    'n_pairs': n,
    'spearman_rho': rho,
    'p_approx': p_approx,
    'p_permutation': p_shuffle,
    'followup_months_min': paired['lag_months_y1'].min(),
    'followup_months_median': paired['lag_months_y1'].median(),
    'followup_months_max': paired['lag_months_y1'].max()
})
paired.to_csv('changes_year1.csv', index=False)
```

### Graph testing if fat fraction and clinical severity change together from baseline to year 1

```python
TEAL = "#1f7a8c"

plt.figure(figsize=(8, 5.5))
plt.scatter(paired['ff_change'], paired['severity_change'],
            color=TEAL, s=70, alpha=0.75)
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.axvline(0, color='gray', linestyle='--', linewidth=1)
plt.title(f'Do fat fraction and clinical severity change together?\n'
          f'Baseline to Year 1 | {n} participants')
plt.xlabel('Change in right TA fat fraction (percentage points)')
plt.ylabel('Change in clinical severity score (points)')
plt.text(0.03, 0.97, f'Spearman rho = {rho:.2f}\nPermutation p = {p_shuffle:.3f}',
         transform=plt.gca().transAxes, va='top',
         bbox=dict(facecolor='white', alpha=0.85, edgecolor='lightgray'))
plt.grid(alpha=0.2)
plt.figtext(0.5, 0.01, 'Change = follow-up minus baseline. Each dot represents one participant; dots may overlap.',
            ha='center', fontsize=8)
plt.tight_layout(rect=[0, 0.04, 1, 1])
plt.show()
```

## Limitations and Future Work:

The dataset had a small sample of 34 participants, so it may not be able to properly reflect the entire population of individuals with FSHD. Also, it is difficult to get people to come back at the exact 1 or 2 year mark, so we had to interpolate the data to make a guess of where their data values would be between each point. With three time points for the main dataset and two timepoints for the MRI samples, it's difficult to truly understand the progression of the disease. Additionally, the MRI information came from one patient, so it cannot necessarily be used to predict the images of other people. Segmentation by computers can also include errors in boundaries that might require human correction. Variability between biomarker levels and functional can also cause issues when analyzing data and they do not perfectly depict the development of FSHD as it could be an off day for the person. Methods like interpolation can also cause data points to be incorrect from the true value at the time interpolated.

If we had more data points, we could be more certain in our analysis because it would be reflective of a larger sample of people. More timepoints would also be helpful so we wouldn't have to make as many assumptions of the year or more between each visit. We also could have improved deep learning models that would be able to segment the MRI data more accurately. Additionally, having more MRI data for different types of muscles could also be useful in ensuring that our analysis is representing the whole body of the FSHD affected individual. With more time, we would compare using the fat fraction of the left side of the body too to make sure that the body is being equally tested.
